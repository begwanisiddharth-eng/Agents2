```python
from openai import OpenAI
openAI = OpenAI()
```

```python
# First call to OpenAI

Messages = [{"role":"user", "content":"Tell me a fun fact"}]
Response = openAI.chat.completions.create(messages=Messages, model="gpt-4o-mini")
Reply = Response.choices[0].message.content
print (Reply)
```

```python
##Question-Answer

Prompt = "Please propose a hard, challenging question to assess someone's IQ. Respond with the question in plain text"
Messages = [{"role":"user", "content":Prompt}]
Response = openAI.chat.completions.create(model="gpt-5.4-mini", messages=Messages)
Question = Response.choices[0].message.content
print (Question + "\n\n")

AnswerResponse = openAI.chat.completions.create(model="gpt-5.4-mini",
                                                messages=[{"role":"user", "content":Question}])
Answer = AnswerResponse.choices[0].message.content
from IPython.display import display, Markdown
display(Markdown(Answer))
```

```python
## Evaluator

Message = f"""
           You are evaluating the response to a question.
           Here is the question:
           {Question}
           And here is the response to the question:
           {Answer}

           Please provide a detailed evaluation of the answer in markdown format
           """

Messages = [{"role":"user", "content":Message}]
Response = openAI.chat.completions.create(model="gpt-5.4", messages=Messages)
display(Markdown(Response.choices[0].message.content))
```

```python
# Finding and solving a pain point in an industry

Prompt =  """
           Choose a business domain (e.g. Finance, Pharma, Travel & Hospitality, Entertainment)
           and within the business domain, choose a problem or a pain point that can be addressed
           via an Agentic AI solution.
           Respond with a clear description of the chosen domain and a detailed description of
           the problem or pain point. Please respond in plain text, without any markdown
           """
Message = [{"role":"user", "content":Prompt}]
Response = openAI.chat.completions.create(model="gpt-4o-mini", messages=Message)
ProblemStatement = Response.choices[0].message.content

Prompt = f"""
           You are provided with a specific industry and a problem area
           within that industry which can be solved with an Agentic AI solution.
           Please provide a detailed solution in markdown format.

           Here is the problem statement which has the industry mentioned and the
           problem area defined in detail:
           {ProblemStatement}
           Start your answer with a brief description of the problem statement and
           then go on to the solution.
           """
Message = [{"role":"user", "content":Prompt}]
Response = openAI.chat.completions.create(model="gpt-5.4-mini", messages=Message)
display(Markdown(Response.choices[0].message.content))
```
